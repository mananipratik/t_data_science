```python
#mean with pandas
import pandas as pd

df = pd.DataFrame({
    'Age': [25, 30, None, 40, 35],
    'Experience': [2, 5, 3, None, 7]
})

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Experience'] = df['Experience'].fillna(df['Experience'].mean())

print(df)
```

```python
#with imputer mean
from sklearn.impute import SimpleImputer
import pandas as pd

df = pd.DataFrame({
    'Age': [25, 30, None, 40, 35],
    'Experience': [2, 5, 3, None, 7]
})

imputer = SimpleImputer(strategy='mean')
df_imputed = imputer.fit_transform(df)

print(df_imputed)
```

```python
#median with pandas
import pandas as pd

df = pd.DataFrame({
    'Age': [25, 30, None, 40, 100],
    'Experience': [2, 5, 3, None, 7]
})

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Experience'] = df['Experience'].fillna(df['Experience'].median())

print(df)
```

```python
#median with imputer
from sklearn.impute import SimpleImputer
import pandas as pd

df = pd.DataFrame({
    'Age': [25, 30, None, 40, 100],
    'Experience': [2, 5, 3, None, 7]
})

imputer = SimpleImputer(strategy='median')
df_imputed = imputer.fit_transform(df)

print(df_imputed)
```

```python
#mode with pandas
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'Finance', None],
    'City': ['Vadodara', 'Ahmedabad', None, 'Vadodara', 'Vadodara']
})

df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
df['City'] = df['City'].fillna(df['City'].mode()[0])

print(df)
```

```python
#mode with imputer
from sklearn.impute import SimpleImputer
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'Finance', None],
    'City': ['Vadodara', 'Ahmedabad', None, 'Vadodara', 'Vadodara']
})

imputer = SimpleImputer(strategy='most_frequent')
df_imputed = imputer.fit_transform(df)

print(df_imputed)
```

```python
#constant with pandas
import pandas as pd

df = pd.DataFrame({
    'Age': [25, None, 30, 40],
    'Department': ['IT', 'HR', None, 'Finance']
})

df['Age'] = df['Age'].fillna(-1)
df['Department'] = df['Department'].fillna('Unknown')

print(df)
```

```python
#constant with imputer
from sklearn.impute import SimpleImputer
import pandas as pd

df = pd.DataFrame({
    'Age': [25, None, 30, 40],
    'Department': ['IT', 'HR', None, 'Finance']
})

num_imputer = SimpleImputer(strategy='constant', fill_value=-1)
cat_imputer = SimpleImputer(strategy='constant', fill_value='Unknown')

df[['Age']] = num_imputer.fit_transform(df[['Age']])
df[['Department']] = cat_imputer.fit_transform(df[['Department']])

print(df)
```

```python
#KNN with imputer
import pandas as pd
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

# Dataset
df = pd.DataFrame({
    'Age': [25, 30, 28, 40, 35],
    'Experience': [2, 5, 4, 8, 6],
    'Salary': [30000, 45000, None, 60000, 55000]
})

# Step 1: Scale data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Step 2: KNN Imputation
imputer = KNNImputer(n_neighbors=2)
df_imputed_scaled = imputer.fit_transform(df_scaled)

# Step 3: Inverse scaling
df_imputed = scaler.inverse_transform(df_imputed_scaled)

print(df_imputed)
```

```python
#TargetEncoder
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [30000, 45000, 35000, 60000, 50000]
})

mean_encoding = df.groupby('Department')['Salary'].mean()

df['Dept_Target_Encoded'] = df['Department'].map(mean_encoding)

print(df)
```

```python
#Frequency
import pandas as pd

df = pd.DataFrame({
    'City': ['Vadodara', 'Ahmedabad', 'Vadodara', 'Surat', 'Vadodara', 'Ahmedabad']
})

freq_map = df['City'].value_counts()

df['City_Freq_Encoded'] = df['City'].map(freq_map)

print(df)
```

```python
#pipeline
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    'Age': [25, 30, None, 40],
    'Department': ['IT', 'HR', 'IT', 'HR'],
    'Salary': [30000, 45000, 35000, 60000]
})

X = df[['Age', 'Department']]
y = df['Salary']

# Separate columns
num_cols = ['Age']
cat_cols = ['Department']

# Pipelines
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='mean'))
])

cat_pipeline = Pipeline([
    ('encoder', OneHotEncoder())
])

# Column Transformer
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

# Final Pipeline
model_pipeline = Pipeline([
    ('preprocessing', preprocessor),
    ('model', LinearRegression())
])

# Train
model_pipeline.fit(X, y)
```
